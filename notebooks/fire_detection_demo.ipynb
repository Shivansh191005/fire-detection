## 3. Load Trained Model

We load the best fine-tuned checkpoint (`best.pt`) if available, or fall back to pretrained `yolov8s.pt`.

```python
from ultralytics import YOLO

weights_path = get_model_path()
print(f"Using model weights: {weights_path}")

model = YOLO(weights_path)
```

## 4. Run Video / Image Inference

Run real-time fire detection on sample videos or images.

```python
# Example video inference
video_path = root_dir / "5622766-uhd_3840_2160_30fps.mp4"

if video_path.exists():
    print(f"Running prediction on {video_path}...")
    results = model.predict(
        source=str(video_path),
        conf=0.35,
        iou=0.70,
        save=True,
        stream=False
    )
    print("Prediction finished! Results saved to runs/detect/")
else:
    print("Sample video not found. Try running on test images directory.")
```

## 5. Model Validation & Performance

Evaluate standard YOLO metrics (`mAP@0.5`, `Precision`, `Recall`) on the validation dataset.

```python
metrics = model.val(data=str(DATA_YAML_PATH), split="val", batch=8)

print("\n--- EVALUATION RESULTS ---")
print(f"mAP50      : {metrics.box.map50:.4f}")
print(f"Precision  : {metrics.box.mp:.4f}")
print(f"Recall     : {metrics.box.mr:.4f}")
```
